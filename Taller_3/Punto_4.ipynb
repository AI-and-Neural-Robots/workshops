**Enunciado:** Genere aleatoriamente una población de $50$ matrices de $120$ por $180$, con números de $0$ a $255$, preséntelas como una gráfica RGB. La función de aptitud es una imagen cualquiera. Evolucione la población inicial hasta llegar a la imagen.

## Análisis del Algoritmo Genético: Evolución de Imágenes RGB

A diferencia de los algoritmos genéticos clásicos que operan sobre vectores binarios o numéricos unidimensionales, este modelo está adaptado para evolucionar **matrices tridimensionales** (imágenes) introduciendo técnicas agresivas para acelerar la convergencia.

A continuación, se detallan las decisiones de diseño fundamentales:

### 1. Representación y Decodificación
*   **Genotipo (Cromosoma):** El cromosoma es una matriz tridimensional (tensor) de dimensiones `120 x 180 x 3` (altura, anchura y canales de color RGB).
*   **Decodificación:** No existe un proceso de decodificación matemático. El genotipo equivale directamente al fenotipo (la imagen visible).

### 2. Función de Aptitud (Fitness)
El objetivo es **minimizar** la diferencia de color respecto a la imagen objetivo.
*   **Métrica:** Se utiliza el **Error Absoluto Medio**.
*   **Cálculo:** Se calcula la diferencia absoluta (`np.abs`) canal por canal (R, G, B) de cada píxel entre la imagen generada y el objetivo. Luego, se promedian (`np.mean`) las 64,800 diferencias individuales de la imagen en un único valor numérico. Entre más cercano a 0 sea este número, mejor es la aptitud.

### 3. Selección y Elitismo
*   **Elitismo Explícito:** Para garantizar que la calidad visual nunca retroceda, los **5 mejores individuos** de cada generación se copian intactos a la siguiente.
*   **Selección de Padres:** Se abandona la selección por ruleta. Se utiliza un método de selección por truncamiento donde los padres se eligen aleatoriamente *exclusivamente* entre los **15 mejores individuos** de la población.

### 4. Operadores Genéticos Guiados
*   **Cruce (Píxel a Píxel):** En lugar de realizar cortes geométricos en la imagen, el cruce es **guiado**. Para cada posición, se evalúa cuál de los dos padres tiene el píxel más cercano al color de la imagen objetivo. El hijo hereda únicamente los "píxeles ganadores", logrando una convergencia acelerada.
*   **Mutación (Perturbación Local o *Jitter*):** Un píxel muta con una probabilidad del 1% ($p_{mut} = 0.01$). La mutación no asigna un color completamente aleatorio, sino que suma o resta un valor pequeño (entre -30 y +30) al color actual. Esto permite refinar gradualmente los tonos sin destruir el progreso del cruce.

```python
# -*- coding: utf-8 -*-
"""Matrices RGB.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/12nofqmQRbgbfROcT2e0FJBJ5x5wJweRZ
"""

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import requests
from io import BytesIO

# 1. Obtener imagen objetivo (120x180 RGB) desde la web
url = "https://picsum.photos/180/120"
response = requests.get(url)
target_img = np.array(Image.open(BytesIO(response.content)).convert('RGB'))

# Configuración del Algoritmo Genético
POP_SIZE = 50
HEIGHT, WIDTH, CHANNELS = 120, 180, 3
GENERATIONS = 50
MUTATION_RATE = 0.01  # Pequeña tasa para perturbar la población

# 2. Generar población inicial aleatoria (50 matrices de 120x180x3)
population = np.random.randint(0, 256, (POP_SIZE, HEIGHT, WIDTH, CHANNELS), dtype=np.uint8)
initial_sample = population[0].copy()

# Lista para guardar la mejor imagen de cada generación
history = []

# Función de aptitud (Error absoluto medio: menor diferencia = mejor aptitud)
def get_fitness(pop):
    return np.mean(np.abs(pop.astype(int) - target_img.astype(int)), axis=(1, 2, 3))

# 3. Proceso evolutivo
for gen in range(GENERATIONS):
    fitness = get_fitness(population)
    idx = np.argsort(fitness)
    population = population[idx]  # Ordenar de mejor a peor

    # Guardar la mejor imagen actual de esta generación
    history.append(population[0].copy())

    # Elitismo: Mantener los 5 mejores individuos
    new_pop = list(population[:5])

    # Reproducción y mutación hasta completar la población
    while len(new_pop) < POP_SIZE:
        # Selección entre los mejores 15 padres
        p1 = population[np.random.randint(0, 15)]
        p2 = population[np.random.randint(0, 15)]

        # Cruce guiado píxel por píxel:
        dist1 = np.abs(p1.astype(int) - target_img.astype(int))
        dist2 = np.abs(p2.astype(int) - target_img.astype(int))

        # Elegimos el píxel del padre que esté más cerca de la imagen objetivo
        child = np.where(dist1 < dist2, p1, p2).astype(np.uint8)

        # Mutación local no guiada (perturbación de valor cercano al actual)
        mut_mask = np.random.rand(HEIGHT, WIDTH, CHANNELS) < MUTATION_RATE
        if np.any(mut_mask):
            # Generamos un cambio aleatorio de entre -10 y +10
            jitter = np.random.randint(-30, 31, size=np.sum(mut_mask))
            mutated_pixels = np.clip(child[mut_mask].astype(int) + jitter, 0, 255).astype(np.uint8)
            child[mut_mask] = mutated_pixels

        new_pop.append(child)

    population = np.array(new_pop)

# 4. Presentación gráfica de todas las generaciones
fig, axes = plt.subplots(1, GENERATIONS + 2, figsize=(3 * (GENERATIONS + 2), 4))

# Mostrar población inicial
axes[0].imshow(initial_sample)
axes[0].set_title("Población Inicial")
axes[0].axis('off')

# Mostrar cada una de las generaciones en su evolución
for i, img in enumerate(history):
    axes[i + 1].imshow(img)
    axes[i + 1].set_title(f"Gen {i + 1}")
    axes[i + 1].axis('off')

# Mostrar imagen objetivo al final
axes[-1].imshow(target_img)
axes[-1].set_title("Objetivo")
axes[-1].axis('off')

plt.tight_layout()
plt.show()
```
